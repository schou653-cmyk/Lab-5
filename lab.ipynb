```python
fighter = Fighter("Aragorn", 100, 8)
mage = Mage("Gandalf", 80, 5)
sword = MeleeWeapon("Sword", 15)
bow = RangedWeapon("Bow", 12,10)
staff = MagicWeapon("Magic", 20,10, 5)
print("characters: ")
print(fighter)
print(mage)
print('\nWeapons: ')
print(sword)
print(bow)
print(staff)

print('\nweapon attributes: ')
print("sword damage: ", sword.damage)
print("bow damage: ", bow.damage)
print("magic staff range: ", staff.range)
print("magic staff mana cost: ", staff.mana_cost)

```
